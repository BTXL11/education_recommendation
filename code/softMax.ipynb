```python
import numpy as np
import math
import torch as th

'''
hardmax
'''
a = np.array([1, 2, 3, 4, 5])
a_max = np.max(a)
print(a_max)

'''
softmax
'''
b = th.FloatTensor([1, 2, 3, 4, 5])
b_max = th.softmax(b, dim=0)
print(b_max)

def q_i_softmax(q, i, T):
    q_exp = th.exp(q / T)
    q_sum = th.sum(q_exp, dim=0)
    q_softmax = th.exp(q[i] / T) / q_sum
    return q_softmax

q = th.FloatTensor([1, 2, 3, 4, 5])
i = 2

print(q_i_softmax(q, i, T=1.0))
```
